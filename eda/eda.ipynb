# EDA — Customer Support Ticket Dataset

Este notebook contém a Análise Exploratória de Dados (EDA) inicial do dataset
que servirá de base para o sistema de atendimento ao cliente construído ao
longo do bloco (classificação de intenção de tickets de suporte).

## Documentação técnica do dataset

**Fonte (origem):** [Customer Support Ticket Dataset](https://www.kaggle.com/datasets/suraj520/customer-support-ticket-dataset/data),
publicado no Kaggle pelo usuário `suraj520`. É um dataset **sintético**,
gerado programaticamente para simular tickets de um sistema de suporte ao
cliente (não são dados reais de uma empresa). Evidências dessa origem
sintética aparecem ao longo desta EDA (ex.: distribuições quase perfeitamente
balanceadas entre categorias e inconsistências de timestamp, ver seção de
qualidade dos dados).

**Licença:** [CC0: Public Domain](https://creativecommons.org/publicdomain/zero/1.0/)
(conforme indicado na página do dataset no Kaggle). Isso significa que o
dataset está em domínio público — pode ser usado, modificado e redistribuído
livremente, sem exigência de atribuição, o que respalda seu uso e
redistribuição neste repositório (arquivo em `data/customer_support_tickets.csv`).

**Principais características:** 8.469 tickets (linhas) e 17 colunas, sem
duplicatas. O dicionário abaixo descreve o significado das colunas mais
relevantes para o problema de classificação de intenção:

| Coluna | O que significa | Valores |
|---|---|---|
| `Ticket ID` | Identificador único do ticket. | 8.469 valores distintos |
| `Ticket Type` | **Categoria da solicitação** — a "intenção" declarada do cliente. É o rótulo candidato do classificador. | 5: Refund request, Technical issue, Cancellation request, Product inquiry, Billing inquiry |
| `Ticket Subject` | **Assunto do ticket**, mais específico que o tipo. Segundo rótulo candidato. | 16: Software bug, Payment issue, Account access, Battery life, Refund request... |
| `Ticket Description` | **Texto livre escrito pelo cliente** — é a principal feature, a única informação disponível quando o ticket chega. | 8.077 textos distintos, ~290 caracteres em média |
| `Ticket Status` | Estágio do atendimento. Determina quais colunas de desfecho estão preenchidas. | 3: Open, Pending Customer Response, Closed |
| `Ticket Priority` | Urgência atribuída ao ticket. | 4: Low, Medium, High, Critical |
| `Ticket Channel` | Canal por onde o ticket entrou. | 4: Email, Phone, Chat, Social media |
| `Product Purchased` | Produto a que o ticket se refere. | 42 produtos |
| `Resolution` | Texto da solução dada. **Só existe para tickets `Closed`.** | 2.769 preenchidos |
| `First Response Time` / `Time to Resolution` | Timestamps de SLA. Preenchidos conforme o status. | ver seção 3 |
| `Customer Satisfaction Rating` | Nota de 1 a 5 dada pelo cliente. **Só existe para tickets `Closed`.** | 2.769 preenchidos |
| `Customer Name` / `Email` / `Age` / `Gender` | Perfil do cliente. Dados pessoais — relevantes para a discussão de confidencialidade na API. | idade de 18 a 70 |

**Relação entre `Ticket Type` e `Ticket Subject`:** os dois campos descrevem a
intenção em níveis diferentes de granularidade, e o esperado seria uma relação
hierárquica — um assunto como *Refund request* deveria pertencer ao tipo
*Refund request*, e *Software bug* ao tipo *Technical issue*. **Não é o que
acontece neste dataset.** Cada um dos 16 assuntos aparece distribuído por todos
os 5 tipos, em proporções praticamente iguais (demonstrado na seção 2.2). Ou
seja, os dois rótulos são atribuídos de forma independente, e não aninhada.

Isso tem consequência direta para a modelagem: os rótulos são mutuamente
inconsistentes (existem tickets com assunto *Refund request* classificados como
tipo *Technical issue*), e só um deles pode ser adotado como alvo do
classificador. Este trabalho adota `Ticket Type` como rótulo principal, por ter
menos classes e corresponder ao roteamento por fila de atendimento.

**Motivo da escolha do dataset:**
1. O texto livre (`Ticket Description`) combinado com campos categóricos
   estruturados (`Ticket Type`, `Ticket Subject`) é exatamente o formato de
   entrada que o futuro agente de atendimento (classificador de intenção)
   precisará processar.
2. O tamanho (aprox. 8.5 mil linhas) é grande o suficiente para treinar um
   classificador de texto simples, mas pequeno o suficiente para rodar EDA e
   experimentos rapidamente em uma máquina local.
3. O dataset tem problemas de qualidade realistas (valores ausentes
   estruturais, inconsistências de timestamp, placeholders de template não
   substituídos), é ótimo para praticar limpeza e validação de dados antes de
   qualquer modelagem, e também para discutir os riscos de treinar um modelo
   sobre dados sintéticos e desbalanceamentos artificiais, que é quando o
   desenvolvedor determina proporções específicas nos dados antes de treinar
   o modelo.
4. Os campos de canal, prioridade e status permitem, no futuro, enriquecer o
   agente com regras de roteamento/priorização além da simples classificação
   de intenção.

## 1. Compreensão do problema e do dataset

**Problema de negócio:** uma equipe de atendimento ao cliente recebe tickets
de suporte por múltiplos canais (email, telefone, chat, redes sociais) e
precisa identificar rapidamente a **intenção** por trás de cada ticket
(ex.: pedido de reembolso, dúvida técnica, cancelamento) para roteá-lo à
fila correta e priorizá-lo. O objetivo final do bloco é construir uma API
com um modelo que, dado o texto de um ticket, prevê essa intenção
automaticamente.

**Papel de cada coluna nesse problema:**
- **Feature candidata principal:** `Ticket Description` (texto livre do
  cliente), é o que estará disponível no momento da predição.
- **Rótulo candidato:** `Ticket Type` e/ou `Ticket Subject`, representam a
  "intenção" que o modelo deverá aprender a prever.
- **Metadados auxiliares:** `Product Purchased`, `Ticket Channel`,
  `Customer Age`/`Gender` podem virar features adicionais.
- **Vazamento de informação (data leakage) a evitar no futuro:** `Resolution`,
  `Time to Resolution` e `Customer Satisfaction Rating` só existem **depois**
  que o ticket é resolvido, não estarão disponíveis no momento em que a
  intenção precisa ser prevista, então não podem ser usadas como features de
  um classificador de intenção.

### 2.1 Classificação formal das variáveis

Antes de qualquer análise, as 17 colunas são classificadas por natureza, porque
é isso que define a técnica aplicável a cada uma (histograma e medidas de
dispersão para numéricas, tabela de frequência para categóricas, estatísticas de
comprimento e vocabulário para textuais, e ordenação/intervalos para temporais).

| Natureza | Colunas | Tratamento nesta EDA |
|---|---|---|
| **Numéricas** | `Customer Age` (discreta, 18–70), `Customer Satisfaction Rating` (ordinal, 1–5) | centro, dispersão e histograma (seção 5.1) |
| **Categóricas** | `Ticket Type`, `Ticket Subject`, `Ticket Status`, `Ticket Priority`, `Ticket Channel`, `Customer Gender`, `Product Purchased` | frequência absoluta e relativa completas (seção 5.2) |
| **Textuais** | `Ticket Description`, `Resolution` | comprimento, unicidade e detecção de boilerplate (seções 3 e 5) |
| **Temporais** | `Date of Purchase`, `First Response Time`, `Time to Resolution` | conversão para datetime e verificação de consistência causal (seção 3) |
| **Identificador / dados pessoais** | `Ticket ID`, `Customer Name`, `Customer Email` | não entram na análise estatística; `Name`/`Email` são dados pessoais e não serão usados como features |

`Ticket Priority` e `Customer Satisfaction Rating` são **ordinais** (têm ordem
natural), o que mais adiante justifica o uso de testes não paramétricos em vez
de testes que pressupõem distribuição normal.

A célula abaixo monta essa classificação a partir do próprio dataframe, para
que ela não dependa de uma lista escrita à mão.

```python
VARIAVEIS = {
    "numerica": ["Customer Age", "Customer Satisfaction Rating"],
    "categorica": ["Ticket Type", "Ticket Subject", "Ticket Status", "Ticket Priority",
                   "Ticket Channel", "Customer Gender", "Product Purchased"],
    "textual": ["Ticket Description", "Resolution"],
    "temporal": ["Date of Purchase", "First Response Time", "Time to Resolution"],
    "identificador/pessoal": ["Ticket ID", "Customer Name", "Customer Email"],
}

classificacao = pd.DataFrame(
    [
        {
            "coluna": col,
            "natureza": natureza,
            "dtype original": str(df[col].dtype),
            "valores distintos": df[col].nunique(),
            "nulos": int(df[col].isnull().sum()),
        }
        for natureza, cols in VARIAVEIS.items()
        for col in cols
    ]
)

# Garante que nenhuma das 17 colunas ficou de fora da classificação
assert set(classificacao["coluna"]) == set(df.columns), "coluna não classificada!"
classificacao
```

### 2.2 Relação entre `Ticket Type` e `Ticket Subject`

Os dois campos são candidatos a rótulo do classificador de intenção, então
precisamos saber se são redundantes, hierárquicos ou independentes.

```python
relacao = pd.crosstab(df["Ticket Subject"], df["Ticket Type"])
print("Assuntos distintos em cada tipo de ticket:")
print(df.groupby("Ticket Type")["Ticket Subject"].nunique())
print(f"\nTotal de assuntos no dataset: {df['Ticket Subject'].nunique()}")
print("\nContagem cruzada (linhas = assunto, colunas = tipo):")
relacao
```

```python
# Proporção de cada tipo dentro de cada assunto (%).
# Se os campos fossem hierárquicos, cada linha teria ~100% concentrado em uma
# única coluna. Valores próximos de 20% em todas as colunas indicam o contrário.
(relacao.div(relacao.sum(axis=1), axis=0) * 100).round(1)
```

**Observações (inspeção inicial):**

1. **Dimensão e tipos:** 8.469 linhas × 17 colunas. O `df.info()` mostra que o
   pandas leu como `object` inclusive as três colunas temporais — elas são
   convertidas para `datetime` na seção 4.
2. **Cardinalidade coerente com a natureza de cada variável:** `Ticket ID` é
   único por linha (identificador); as categóricas têm de 3 a 42 valores; as
   textuais têm cardinalidade quase igual ao número de linhas.
3. **`Customer Email` tem 8.320 valores distintos para 8.469 linhas**, ou seja,
   alguns clientes abriram mais de um ticket. Não é duplicata de registro.
4. **Nulos concentrados em 3 colunas**, todas ligadas ao desfecho do ticket —
   investigado na seção 3, onde se confirma que a ausência é estrutural.
5. **`Ticket Type` e `Ticket Subject` são independentes, não hierárquicos:**
   os 16 assuntos aparecem nos 5 tipos, e dentro de cada assunto os tipos se
   dividem em proporções próximas de 20%. Um assunto *Refund request* tem
   praticamente a mesma chance de estar marcado como *Technical issue* ou como
   *Refund request*. É um defeito de rotulagem do dataset sintético, que impede
   usar os dois campos como uma hierarquia de intenção e vira a hipótese **H3**
   da seção 6.

## 3. Verificação da qualidade dos dados

```python
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(1)
pd.DataFrame({"missing_count": missing, "missing_pct": missing_pct}).query("missing_count > 0")
```

As três colunas com valores ausentes (`Resolution`, `Time to Resolution`,
`Customer Satisfaction Rating`) têm **exatamente 5.700 valores nulos** cada.
A hipótese é que a ausência não é aleatória (MNAR — *missing not at random*),
e sim **estrutural**: essas colunas só fazem sentido para tickets já
resolvidos. A célula abaixo confirma isso cruzando com `Ticket Status`.

```python
print("Nulos em Resolution por status:")
print(pd.crosstab(df["Ticket Status"], df["Resolution"].isnull()))
print()
print("Nulos em First Response Time por status:")
print(pd.crosstab(df["Ticket Status"], df["First Response Time"].isnull()))
```

Como podemos ver: `Resolution`/`Time to Resolution`/`Customer Satisfaction Rating`
são nulos em **100% dos tickets não-`Closed`** e preenchidos em 100% dos
`Closed`. Da mesma forma, `First Response Time` só é nulo quando o status é
`Open` (ainda não houve nenhuma resposta). Ou seja: não é um problema de
qualidade a ser corrigido, é o comportamento esperado do domínio,
a limpeza deve **preservar** esses nulos e, no máximo, derivar uma flag
booleana a partir deles.

### 3.1 Consistência de escrita das categorias e dos textos

Categorias "iguais" escritas de formas diferentes (`Email` vs `email` vs
`Email `) seriam contadas como classes distintas e inflariam artificialmente a
cardinalidade. O mesmo vale para textos vazios ou compostos só de espaços, que
passam despercebidos por `isnull()`.

```python
texto_e_categoria = (
    VARIAVEIS["categorica"] + VARIAVEIS["textual"] + ["Customer Name", "Customer Email"]
)

relatorio = []
for col in texto_e_categoria:
    serie = df[col].dropna().astype(str)
    relatorio.append({
        "coluna": col,
        "valores distintos": serie.nunique(),
        # se normalizar (minúsculas + sem espaços nas pontas) reduz a contagem,
        # é porque existem variantes de escrita do mesmo valor
        "variantes de escrita": serie.nunique() - serie.str.strip().str.lower().nunique(),
        "com espaço extra nas pontas": int((serie != serie.str.strip()).sum()),
        "vazios / só espaços": int((serie.str.strip() == "").sum()),
    })

pd.DataFrame(relatorio).set_index("coluna")
```

**Resultado:** nenhuma coluna categórica apresenta variantes de escrita,
espaços extras nas pontas ou valores vazios — as categorias já chegam
padronizadas, e nenhuma normalização de texto é necessária antes da análise.

A única divergência aparece em `Ticket Description`: duas descrições que diferem
apenas por maiúsculas/minúsculas. Como é texto livre, isso não é um problema de
padronização de categoria e não exige correção.

Essa regularidade total é, mais uma vez, indício de geração sintética: em dados
reais de suporte, campos preenchidos por operadores costumam ter variantes de
escrita.

**Resumo dos problemas de qualidade encontrados:**
1. Sem duplicatas de `Ticket ID` e sem linhas 100% duplicadas — ok.
2. Nulos estruturais (não é erro) em `Resolution`, `Time to Resolution` e
   `Customer Satisfaction Rating` para tickets ainda não `Closed`.
3. **Bug de template:** 100% das descrições contêm o placeholder literal
   `{product_purchased}` não substituído pelo nome real do produto — um
   defeito de geração de texto no dataset sintético.
4. **Inconsistência temporal:** uma fração relevante dos tickets fechados
   tem `Time to Resolution` **anterior** a `First Response Time`, o que é
   logicamente impossível em um fluxo real de atendimento — evidência de que
   os timestamps foram gerados aleatoriamente, sem garantir ordem causal.
   Por isso, `First Response Time`/`Time to Resolution` **não devem ser
   usados** para calcular SLAs reais nesta EDA. Ficam registrados aqui como
   limitação do dataset.


## 4. Limpeza e preparação dos dados

Com base no diagnóstico acima, serão feitas apenas conversões de tipo e a criação de colunas derivadas úteis
para a análise univariada e para etapas futuras de modelagem. **Nenhuma linha é removida** e **nenhum campo nulo é
alterado** (imputar inventaria uma resolução/satisfação que não existe).


```python
df_clean = df.copy()

# 4.1 Tipos de data/hora
df_clean["Date of Purchase"] = pd.to_datetime(df_clean["Date of Purchase"])
df_clean["First Response Time"] = pd.to_datetime(df_clean["First Response Time"])
df_clean["Time to Resolution"] = pd.to_datetime(df_clean["Time to Resolution"])

# 4.2 Categóricas como category (mais eficiente e explícito)
for col in ["Ticket Type", "Ticket Subject", "Ticket Status", "Ticket Priority",
            "Ticket Channel", "Customer Gender"]:
    df_clean[col] = df_clean[col].astype("category")

# 4.3 Flags derivadas a partir dos nulos estruturais (em vez de imputar)
df_clean["is_closed"] = df_clean["Ticket Status"].eq("Closed")
df_clean["has_satisfaction_rating"] = df_clean["Customer Satisfaction Rating"].notna()

# 4.4 Remove o placeholder de template não substituído para uma versão "limpa" do texto
df_clean["description_clean"] = (
    df_clean["Ticket Description"].str.replace(r"\{product_purchased\}", "the product", regex=True)
)
df_clean["description_length"] = df_clean["Ticket Description"].str.len()

# 4.5 Ano/mês da compra, útil para eventuais análises temporais futuras
df_clean["purchase_year"] = df_clean["Date of Purchase"].dt.year
df_clean["purchase_month"] = df_clean["Date of Purchase"].dt.month

df_clean.dtypes
```

```python
# Checagem pós-limpeza: mesma contagem de linhas, nulos estruturais preservados
assert len(df_clean) == len(df)
assert df_clean["Customer Satisfaction Rating"].isnull().sum() == 5700
df_clean[["Ticket Status", "is_closed", "has_satisfaction_rating",
          "description_length", "purchase_year"]].head()
```

## 5. Análise Univariada

Distribuição das principais variáveis categóricas e numéricas do dataset.

### 5.1 Variáveis numéricas

Para cada variável numérica: (1) estatísticas de centro (média, mediana, mínimo,
máximo), (2) medidas de dispersão (quartis, variância, desvio padrão) e (3) a
distribuição via histograma.

```python
numeric_cols = ["Customer Age", "Customer Satisfaction Rating"]

for col in numeric_cols:
    serie = df_clean[col].dropna()
    print(f"--- {col} ---")
    print(f"média: {serie.mean():.2f} | mediana: {serie.median():.2f} | "
          f"mínimo: {serie.min():.2f} | máximo: {serie.max():.2f}")
    q1, q2, q3 = serie.quantile([0.25, 0.5, 0.75])
    print(f"Q1: {q1:.2f} | Q2 (mediana): {q2:.2f} | Q3: {q3:.2f}")
    print(f"variância: {serie.var():.2f} | desvio padrão: {serie.std():.2f}")
    print()
```

**Observações (variáveis numéricas):**
- `Customer Age`: média (~44) e mediana (~44) praticamente coincidem, e o
  desvio padrão (~15.3) é compatível com um intervalo uniforme de 18 a 70 anos,
  não há assimetria nem indício de outliers de idade.
- `Customer Satisfaction Rating` (calculada apenas sobre os 2.769 tickets
  `Closed`): média e mediana (~3) também coincidem, com desvio padrão baixo
  (~1.4) em uma escala discreta de 1 a 5, distribuição aproximadamente
  simétrica, sem uma nota dominante.
- Em ambos os casos, média ≈ mediana sugere baixa assimetria, reforçando a
  leitura de que o dataset foi gerado sinteticamente com distribuições
  próximas do uniforme, e não amostrado de um cenário real de suporte.

### 5.2 Variáveis categóricas

Para cada variável categórica: (1) quantidade de categorias, (2) frequência
absoluta e relativa (%) e (3) a categoria mais frequente (moda).

```python
categorical_cols = VARIAVEIS["categorica"]

for col in categorical_cols:
    serie = df_clean[col]
    freq_abs = serie.value_counts()
    freq_rel = (serie.value_counts(normalize=True) * 100).round(2)
    moda = serie.mode()[0]
    print(f"--- {col} ---")
    print(f"quantidade de categorias: {serie.nunique()}")
    print(f"categoria mais frequente (moda): '{moda}' "
          f"({freq_abs[moda]} tickets, {freq_rel[moda]}%)")
    # Frequências COMPLETAS: todas as categorias, não apenas as mais frequentes.
    tabela = pd.DataFrame({"freq. absoluta": freq_abs, "freq. relativa (%)": freq_rel})
    print(tabela.to_string())
    print(f"soma: {freq_abs.sum()} tickets | {freq_rel.sum():.2f}%")
    print()
```

**Observações (variáveis categóricas):**

As tabelas acima trazem a frequência completa de cada variável (todas as
categorias, não só as mais frequentes), o que permite verificar que não existem
categorias raras escondidas na cauda:

- `Ticket Type` (5 categorias): varia de 1.634 (Billing inquiry, 19,29%) a
  1.752 tickets (Refund request, 20,69%) — amplitude de apenas 1,4 ponto
  percentual entre a categoria mais e a menos frequente.
- `Ticket Priority` (4 categorias) e `Ticket Channel` (4 categorias): todas
  próximas de 25%. `Customer Gender` (3 categorias): todas próximas de 33%.
- `Ticket Subject` (16 categorias): todas entre 5,64% e 6,80%, sem nenhuma
  categoria abaixo de 1% — nenhuma classe rara.
- `Ticket Status` (3 categorias): Pending Customer Response 34,0%, Open 33,3%,
  Closed 32,7%.
- `Product Purchased` (42 categorias): a mais frequente representa apenas ~2,8%
  do total; nenhuma concentração relevante em um produto específico.

Esse padrão de frequências quase uniformes em **todas** as variáveis
categóricas, sem cauda longa e sem categoria dominante, é o principal indício
estatístico de que o dataset é sintético — e sustenta a hipótese H1 da seção 6.

## 6. Hipóteses sobre as "intenções" dos usuários

A partir dos padrões observados na análise univariada, formulamos quatro
hipóteses sobre como a intenção do usuário se manifesta neste dataset. Cada uma
é enunciada de forma **verificável**: traz a evidência observada na EDA e o
teste estatístico que a confirma ou refuta no TP2.

Em todas elas, "intenção" é operacionalizada como a coluna `Ticket Type`, pelos
motivos discutidos na seção 2.2.

---

**H1 — A distribuição das intenções é artificialmente uniforme.**

*Enunciado:* as cinco intenções ocorrem em proporções estatisticamente
indistinguíveis de 20% cada, ao contrário de um fluxo real de suporte, em que
problemas técnicos e cobrança costumam dominar.

*Evidência na EDA:* seção 5.2 — as frequências vão de 19,29% (Billing inquiry,
1.634) a 20,69% (Refund request, 1.752).

*Como verificar:* teste qui-quadrado de aderência contra a distribuição
uniforme. Não rejeitar H₀ sustenta a hipótese.

*Por que importa:* um classificador treinado nessa distribuição aprende um
*prior* irreal. Em produção, com intenções desbalanceadas, a acurácia global cai
e a métrica honesta passa a ser F1 por classe, não acurácia.

---

**H2 — As intenções se organizam em dois eixos: "algo está quebrado" e "algo na
conta/pedido precisa mudar".**

*Enunciado:* as 16 categorias de `Ticket Subject` se agrupam em dois blocos
semanticamente coerentes — técnico (Software bug, Hardware issue, Network
problem, Battery life, Display issue, Data loss, Installation support, Product
setup, Product compatibility, Peripheral compatibility) e comercial/conta
(Refund request, Payment issue, Account access, Cancellation request, Delivery
problem, Product recommendation).

*Evidência na EDA:* seção 5.2 — a leitura das 16 categorias mostra os dois
blocos; nenhum assunto fica fora deles.

*Como verificar:* criar a variável derivada `eixo` (técnico/comercial) e
comparar os dois grupos em uma variável numérica — por exemplo a satisfação do
cliente nos tickets `Closed` — com o teste de Mann-Whitney (não paramétrico,
porque a nota é ordinal de 1 a 5).

*Por que importa:* se o agrupamento tiver sinal, um classificador hierárquico
(primeiro o eixo, depois o assunto) tende a ser mais robusto do que prever 16
classes de uma vez.

---

**H3 — `Ticket Type` e `Ticket Subject` são rótulos independentes, e não uma
hierarquia de intenção.**

*Enunciado:* o assunto do ticket não determina — nem sequer informa — o tipo
registrado. Um ticket com assunto *Refund request* tem a mesma probabilidade de
estar marcado como *Technical issue* ou como *Refund request*.

*Evidência na EDA:* seção 2.2 — os 16 assuntos aparecem nos 5 tipos, e dentro de
cada assunto os tipos se distribuem em proporções próximas de 20%.

*Como verificar:* teste qui-quadrado de independência na tabela de contingência
`Ticket Subject` × `Ticket Type`. Um p-valor alto (não rejeitar a independência)
confirma a hipótese.

*Por que importa:* é a hipótese de maior impacto prático. Se confirmada, pelo
menos um dos dois campos é ruído de rotulagem, e treinar o classificador para
prever `Ticket Subject` a partir do texto estaria tentando aprender uma relação
que não existe nos dados. Define qual coluna pode ser usada como alvo.

---

**H4 — Nem os metadados estruturados nem o tamanho do texto carregam sinal sobre
a intenção.**

*Enunciado:* `Ticket Channel`, `Ticket Priority`, `Customer Gender` e o
comprimento de `Ticket Description` são independentes de `Ticket Type`, de modo
que nenhum deles serve como atalho para prever a intenção.

*Evidência na EDA:* seção 5 — todas essas variáveis têm distribuição
praticamente uniforme, e o comprimento médio das descrições fica em torno de 290
caracteres para qualquer tipo de ticket. Além disso, 100% das descrições contêm
o placeholder `{product_purchased}` não substituído e frases repetidas de
template ("Please assist.", "I've tried troubleshooting steps mentioned in the
user manual...").

*Como verificar:* qui-quadrado de independência para cada variável categórica
contra `Ticket Type`, e Kruskal-Wallis (ou Mann-Whitney entre dois eixos) para o
comprimento da descrição.

*Por que importa:* se confirmada, o único sinal disponível está no conteúdo
semântico do texto livre, não em features baratas de metadado — o que define o
tipo de modelo a construir (representação de texto) e antecipa um teto baixo de
desempenho neste dataset sintético.

### 7.1 Preparação das variáveis numéricas

O dataset traz apenas duas colunas numéricas de origem (`Customer Age` e
`Customer Satisfaction Rating`), o que daria uma matriz de correlação de 2×2 —
pouco informativa. Antes do heatmap, montamos um conjunto numérico maior,
aproveitando as colunas derivadas criadas na seção 4 e codificando a prioridade,
que é **ordinal** (Low < Medium < High < Critical) e por isso pode virar número
sem distorcer a leitura.

A diferença em horas entre a compra e a primeira resposta entra com uma
ressalva: a seção 3 mostrou que os timestamps do dataset são causalmente
inconsistentes, então essa coluna serve para a análise de correlação, mas **não**
para afirmar qualquer coisa sobre SLA real.

```python
ORDEM_PRIORIDADE = {"Low": 1, "Medium": 2, "High": 3, "Critical": 4}

df_num = pd.DataFrame({
    "idade": df_clean["Customer Age"],
    "satisfacao": df_clean["Customer Satisfaction Rating"],
    "tamanho_descricao": df_clean["description_length"],
    "ano_compra": df_clean["purchase_year"],
    "mes_compra": df_clean["purchase_month"],
    "prioridade_ordinal": df_clean["Ticket Priority"].map(ORDEM_PRIORIDADE).astype("float"),
    "ticket_fechado": df_clean["is_closed"].astype(int),
    "horas_ate_primeira_resposta": (
        df_clean["First Response Time"] - df_clean["Date of Purchase"]
    ).dt.total_seconds() / 3600,
})

print("Variáveis numéricas disponíveis para a correlação:", len(df_num.columns))
df_num.describe().round(2)
```

### 7.2 Heatmap de correlação

Correlação de **Spearman**, não de Pearson: parte das variáveis é ordinal
(`satisfacao`, `prioridade_ordinal`) ou binária (`ticket_fechado`), e Spearman
mede relação monotônica sem exigir escala de intervalo nem normalidade.

```python
correlacao = df_num.corr(method="spearman")

fig, ax = plt.subplots(figsize=(9, 7))
sns.heatmap(
    correlacao,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    square=True,
    linewidths=0.5,
    cbar_kws={"label": "correlação de Spearman"},
    ax=ax,
)
ax.set_title("Heatmap de correlação entre as variáveis numéricas")
plt.tight_layout()
plt.show()

# Maiores correlações em módulo, ignorando a diagonal
pares = (
    correlacao.where(~np.eye(len(correlacao), dtype=bool))
    .stack()
    .abs()
    .sort_values(ascending=False)
)
print("Maiores correlações em módulo:")
print(pares.head(6).round(3).to_string())
```

**Leitura do heatmap.** Duas correlações saltam à vista, e **nenhuma das duas
diz algo sobre atendimento**: `horas_ate_primeira_resposta` aparece correlacionada
com `ano_compra` (0,87) e com `mes_compra` (0,51). Isso é artefato de construção —
a coluna foi calculada como a diferença entre a primeira resposta e a data da
compra, então ela carrega, por definição, a variação da própria data de compra.
Serve de lembrete metodológico: variável derivada sempre aparece correlacionada
com a variável que a originou, e interpretar isso como achado seria um erro.

Descontados esses pares, **a maior correlação de todo o conjunto é 0,04**
(satisfação × horas até a primeira resposta) — ou seja, zero na prática. Idade,
prioridade, tamanho da descrição e satisfação não se relacionam entre si de
nenhuma forma detectável.

Em dados reais esperaríamos o oposto: prioridade correlacionada com tempo de
resposta, e satisfação caindo conforme o atendimento demora. A matriz quase toda
branca é a confirmação visual de que as variáveis foram geradas de forma
independente umas das outras — a mesma evidência que as hipóteses H1 e H4
apontam, agora em forma de matriz.

### 7.4 Teste de hipótese formal — H2, com Mann-Whitney

**Hipótese testada (H2 da seção 6):** as intenções se organizam em dois eixos,
"algo está quebrado" (técnico) e "algo na conta/pedido precisa mudar"
(comercial). Se esse agrupamento tiver significado real, os dois grupos devem se
comportar de forma diferente em alguma medida de desfecho — a mais direta é a
satisfação do cliente.

- **H₀ (hipótese nula):** a distribuição da satisfação é a mesma nos dois eixos.
- **H₁ (alternativa):** as distribuições diferem.
- **Teste:** Mann-Whitney U, bilateral, com α = 0,05.
- **Por que Mann-Whitney e não t-test:** a satisfação é uma nota **ordinal** de 1
  a 5, discreta e sem distribuição normal. O t-test pressupõe normalidade e
  trataria a diferença entre as notas 1 e 2 como igual à diferença entre 4 e 5.
  Mann-Whitney compara postos (ordem), que é o que uma escala ordinal permite.
- **Amostra:** apenas os tickets `Closed`, os únicos com nota de satisfação.

```python
tecnicos = fechados.loc[fechados["eixo"] == "técnico", "Customer Satisfaction Rating"].dropna()
comerciais = fechados.loc[fechados["eixo"] == "comercial", "Customer Satisfaction Rating"].dropna()

print(f"n (técnico)  = {len(tecnicos)}  | mediana = {tecnicos.median():.2f} | média = {tecnicos.mean():.3f}")
print(f"n (comercial)= {len(comerciais)} | mediana = {comerciais.median():.2f} | média = {comerciais.mean():.3f}")

u, p_valor = stats.mannwhitneyu(tecnicos, comerciais, alternative="two-sided")

# Tamanho de efeito (rank-biserial): 0 = nenhuma diferença, ±1 = separação total
rank_biserial = 1 - (2 * u) / (len(tecnicos) * len(comerciais))

print(f"\nMann-Whitney U = {u:.1f}")
print(f"p-valor        = {p_valor:.4f}")
print(f"tamanho de efeito (rank-biserial) = {rank_biserial:.4f}")
print("\nConclusão:", "REJEITA H0 (diferença significativa)" if p_valor < 0.05
      else "NÃO rejeita H0 (sem diferença detectável)")
```

#### Interpretação do p-valor, em linguagem acessível

O p-valor responde a uma pergunta bem específica: **se não houvesse nenhuma
diferença real entre os dois grupos, qual seria a chance de observarmos uma
diferença do tamanho desta, só por sorteio?**

O resultado foi **p ≈ 0,75**, ou seja, cerca de 75% — altíssima. Uma diferença
deste tamanho entre tickets técnicos e comerciais é absolutamente banal em dados
sem estrutura nenhuma. Como é muito maior que o limite de 5% que adotamos,
**não rejeitamos a hipótese nula**: os dados não mostram diferença de satisfação
entre os dois eixos de intenção.

O tamanho de efeito (rank-biserial ≈ −0,007) diz a mesma coisa por outro caminho:
mesmo a diferença observada é praticamente nula. Em termos práticos, pegar um
ticket técnico e um comercial ao acaso dá 49,6% de chance de o técnico ter a
nota maior — o mesmo que jogar uma moeda.

**Cuidado importante:** "não rejeitar H₀" não é o mesmo que "provar que são
iguais". O teste diz que, com estes dados, não há evidência de diferença. Com um
dataset sintético em que todas as variáveis foram geradas independentemente,
este é exatamente o resultado esperado — e reportá-lo é mais honesto do que
procurar um corte de dados que produzisse um p-valor pequeno.

**Conexão com o TP1:** a H2 propunha que o agrupamento em dois eixos seria um
atalho útil para o classificador, via um modelo hierárquico (primeiro o eixo,
depois o assunto). O teste **não sustenta** esse caminho: os eixos não se
distinguem nem na única variável de desfecho disponível. A hipótese foi testada e
o caminho hierárquico fica descartado como atalho — o que é um resultado útil,
porque evita construir um modelo em duas etapas que não teria base nos dados.

### 7.5 Verificação das demais hipóteses

As outras três hipóteses da seção 6 pedem testes de independência e de
comparação entre vários grupos.

```python
print("=" * 70)
print("H1 — a distribuição das intenções é uniforme?  (qui-quadrado de aderência)")
observado = df_clean["Ticket Type"].value_counts()
esperado = [len(df_clean) / len(observado)] * len(observado)
chi2_h1, p_h1 = stats.chisquare(f_obs=observado.values, f_exp=esperado)
print(observado.to_string())
print(f"chi2 = {chi2_h1:.2f} | p = {p_h1:.4f} ->",
      "distribuição compatível com a uniforme" if p_h1 > 0.05 else "distribuição NÃO é uniforme")

print("\n" + "=" * 70)
print("H3 — Ticket Type e Ticket Subject são independentes?  (qui-quadrado de independência)")
tabela_h3 = pd.crosstab(df_clean["Ticket Subject"], df_clean["Ticket Type"])
chi2_h3, p_h3, gl_h3, _ = stats.chi2_contingency(tabela_h3)
print(f"chi2 = {chi2_h3:.2f} | graus de liberdade = {gl_h3} | p = {p_h3:.4f} ->",
      "INDEPENDENTES (não são hierárquicos)" if p_h3 > 0.05 else "existe associação")

print("\n" + "=" * 70)
print("H4 — os metadados estruturados informam a intenção?")
for coluna in ["Ticket Channel", "Ticket Priority", "Customer Gender"]:
    tabela = pd.crosstab(df_clean[coluna], df_clean["Ticket Type"])
    chi2, p, gl, _ = stats.chi2_contingency(tabela)
    print(f"  {coluna:18s} x Ticket Type: chi2 = {chi2:6.2f} | p = {p:.4f} ->",
          "independente" if p > 0.05 else "associado")

grupos = [g["description_length"].values for _, g in df_clean.groupby("Ticket Type", observed=True)]
h_stat, p_kw = stats.kruskal(*grupos)
print(f"  tamanho da descrição entre os 5 tipos (Kruskal-Wallis): H = {h_stat:.2f} | p = {p_kw:.4f} ->",
      "sem diferença" if p_kw > 0.05 else "há diferença")
```

**Resultado consolidado das quatro hipóteses:**

| Hipótese | Teste | Resultado | Veredito |
|---|---|---|---|
| **H1** — distribuição de intenções artificialmente uniforme | Qui-quadrado de aderência | p ≈ 0,11 | **Sustentada.** As cinco intenções são compatíveis com uma distribuição uniforme |
| **H2** — intenções se agrupam em eixo técnico e comercial | Mann-Whitney | p ≈ 0,75 | **Não sustentada.** Os eixos não se distinguem na satisfação |
| **H3** — `Ticket Type` e `Ticket Subject` são independentes | Qui-quadrado de independência | p ≈ 0,98 | **Sustentada.** Os dois rótulos não formam hierarquia |
| **H4** — metadados e tamanho do texto não informam a intenção | Qui-quadrado + Kruskal-Wallis | todos p > 0,05 | **Sustentada.** Nenhum metadado se associa à intenção |

**O que isso significa para o projeto, em uma frase:** o dataset é sintético e
suas variáveis foram geradas de forma independente, então nenhuma feature
estruturada prevê a intenção, e os dois campos candidatos a rótulo se
contradizem — a classificação do TP3 terá de escolher **um** alvo e apostar
exclusivamente no conteúdo semântico do texto livre, com expectativa realista de
desempenho modesto. O detalhamento e os próximos passos estão em
[`docs/relatorio_eda.md`](../docs/relatorio_eda.md).
